## Load libraries

```python
from fly_tracking import *
import multiprocessing as mp
```

## Set parameters

```python
MOVIE_DIR = "./videos"
OUT_DIR = "./output"
OVERWRITE = False
NB_BG_FRAMES = int(1000)
MAX_TIME = int(900)
MAX_HISTORY = int(10)

sub_dir, video_files = scandir_fast(MOVIE_DIR, ".mp4")
```

## Generate background images

```python
pool = mp.Pool(mp.cpu_count())

[
    pool.apply_async(
        create_background_image,
        args=(video, OUT_DIR, NB_BG_FRAMES, MAX_TIME, OVERWRITE),
    )
    for video in video_files
]

pool.close()
pool.join()
```

## Detect moving objects

```python
pool = mp.Pool(mp.cpu_count())

[
    pool.apply_async(detect_moving_objects, args=(video, OUT_DIR, MAX_TIME, OVERWRITE))
    for video in video_files
]

pool.close()
pool.join()
```

## Track moving objects

```python
pool = mp.Pool(mp.cpu_count())

[
    pool.apply_async(track_moving_objects, args=(video, OUT_DIR, MAX_TIME, OVERWRITE))
    for video in video_files
]

pool.close()
pool.join()
```

## Connect tracklets

```python
pool = mp.Pool(mp.cpu_count())

[
    pool.apply_async(connect_tracklets, args=(video, OUT_DIR, MAX_TIME, OVERWRITE))
    for video in video_files
]

pool.close()
pool.join()
```

## Generate annotated detection videos

```python
pool = mp.Pool(mp.cpu_count())

[
    pool.apply_async(save_detection_movie, args=(video, OUT_DIR, MAX_TIME, OVERWRITE))
    for video in video_files
]

pool.close()
pool.join()
```

## Generate annotated tracking videos

```python
pool = mp.Pool(mp.cpu_count())

[
    pool.apply_async(
        save_tracking_movie, args=(video, OUT_DIR, MAX_TIME, MAX_HISTORY, OVERWRITE)
    )
    for video in video_files
]

pool.close()
pool.join()
```

## Generate leftovers videos

```python
pool = mp.Pool(mp.cpu_count())

[
    pool.apply_async(save_leftovers_movie, args=(video, OUT_DIR, MAX_TIME, OVERWRITE))
    for video in video_files
]

pool.close()
pool.join()
```

## Generate long tracks videos

```python
pool = mp.Pool(mp.cpu_count())

[
    pool.apply_async(
        save_longtracks_movie, args=(video, OUT_DIR, MAX_TIME, MAX_HISTORY, OVERWRITE)
    )
    for video in video_files
]

pool.close()
pool.join()
```
